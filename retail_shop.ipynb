```python
expensive_items=df[df['final_price']>20000]
print(expensive_items)
```

```python
result=df.groupby('brand').agg(
    total_products=('brand', 'count'),
    avg_price=('final_price', 'mean')
).reset_index
print(result)
```

```python
filtered_result=(
    df[df['final_price']>2000]
    .groupby('brand')
    .agg(
        total_products=('brand','count'),
        avg_price=('final_price', 'mean')
    )
    .reset_index()
    .sort_values(by='avg_price',ascending=False)
)
print(filtered_result)
```

```python
import numpy as np
df_filtered=df[(df['category']== 'Fashion') & (df['stock_available']>5)].copy()
df_filtered['price_tier']= np.where(df_filtered['final_price']>2000, 'High Price', 'Low Price')

result_tier=(
    df_filtered.groupby(['brand', 'price_tier'])
    .agg(
        total_items=('brand', 'count'),
        avg_price=('final_price', 'mean')
    )
    .reset_index()
    .sort_values(by='total_items', ascending=False)
)
print(result_tier)
```
